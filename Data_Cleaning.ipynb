```python
df_match.rename({"scoreboard":"match_id"},axis=1,inplace=True)
```

```python
match_ids_dict={}
for index ,row in df_match.iterrows():
    key1 = row['team1']+ ' Vs ' +row['team2']
    key2 = row['team2']+ ' Vs ' +row['team1']

    match_ids_dict[key1] = row["match_id"]
    match_ids_dict[key2] = row["match_id"]
```

```python
with open("wc_player_info_complete.json")as f:
    data=json.load(f)
```

```python
new_players = pd.DataFrame([
    ["Simon Ssesazi", "Uganda", "Left hand Bat", "Right arm Medium", "Wicketkeeper Batter",""],
    ["Bilal Hassan", "Uganda", "Right hand Bat", "Right arm Medium fast", "Bowler",""],
    ["Henry Ssenyondo", "Uganda", "Right hand Bat", "Slow left arm orthodox", "Bowler",""],
    ["Karim Janat", "Afghanistan", "Right hand Bat", "Right arm Medium fast", "Allrounder",""],
    ["Mark Chapman", "New Zealand", "Left hand Bat", "Slow left arm orthodox", "Middle order Batter",""],
    ["Michael Bracewell", "New Zealand", "Left hand Bat", "Right arm Offbreak", "Allrounder",""],
    ["Matt Henry", "New Zealand", "Right hand Bat", "Right arm Fast medium", "Bowler",""]
], columns=df_player.columns)

df_player = pd.concat([df_player, new_players], ignore_index=True)
```

```python
len(df_player)
```

```python
df_player

```

```python
print("Total rows:", len(df_player))
print("Unique players:", df_player["name"].nunique())
```

```python
duplicates = df_player[
    df_player["name"].duplicated(keep=False)
].sort_values("name")

print(duplicates[["name", "team"]])
```

```python
df_player = df_player.drop_duplicates(subset="name")
df_player = df_player.reset_index(drop=True)

print("Total players:", len(df_player))
```

```python
df_player.to_csv('DATA/fact_player_summary.csv', index=False)
```
